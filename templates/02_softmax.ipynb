# 🟢 Easy: Implement Softmax

Implement the **Softmax** function from scratch.

$$\text{softmax}(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$$

### Signature
```python
def my_softmax(x: torch.Tensor, dim: int = -1) -> torch.Tensor:
    ...
```

### Rules
- Do **NOT** use `torch.softmax`, `F.softmax`, or `torch.nn.Softmax`
- Must be **numerically stable** (hint: subtract `max` before `exp`)

### Example
```
Input:  tensor([1., 2., 3.])
Output: tensor([0.0900, 0.2447, 0.6652])  # sums to 1.0
```

````python
import torch
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def my_softmax(x: torch.Tensor, dim: int = -1) -> torch.Tensor:
    x_max = x.max(dim=dim, keepdim=True).values # max同时返回最大值和其索引
    # 首先数学上减去max和原始公式等价，数学证明很简单。
    # 其次减去最大值可以防止溢出，例如x=[1000,1001,1002]时，e^1000会溢出，减去最大值保证e^(x-x_max)落在0-1之间
    e_x = torch.exp(x - x_max)
    # sum(dim, keepdim)指对某一个维度求和并且维度数量不变
    # 假如e_x是（2，3）的tensor，e_x.sum(1, False)会按列求和，维度降为1维，shape为(3)，注意此时(2,3)/(3)可以被广播
    # e_x.sum(1, True)依旧会按列求和，但维度数量不变，shape变成（1，3）
    # (2,3) / (1,3) 会被广播为 (2,3)/(2,3)
    # e_x.sum(0, False)会按行求和，维度降为1维，shape为(2)，注意此时(2,3)/（2）无法被广播
    # e_x.sum(0, True)依旧会按行求和，但维度数量不变，shape变成（2，1）
    # (2,3) / (2,1) 会被广播为 (2,3)/(2,3)
    return e_x / e_x.sum(dim=dim, keepdim=True)
````

````python
# 🧪 Debug
x = torch.tensor([1.0, 2.0, 3.0])
print("Output:", my_softmax(x, dim=-1))
print("Sum:   ", my_softmax(x, dim=-1).sum())  # should be ~1.0
print("Ref:   ", torch.softmax(x, dim=-1))
````
